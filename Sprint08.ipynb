Mediante las siguientes funciones se corregirá la codificación del texto del dataframe.
Los errores en acentos y otros signos, como por ejemplo, `Muñoz` que aparece como `MuÃ±oz`. Se deben a que el texto original está en UTF-8, pero se ha interpretado como si estuviera en otra codificación (latin-1 o cp1252).

La primera función revierte ese error: convierte el texto de nuevo a bytes usando la codificación incorrecta y lo decodifica como UTF-8. Si el texto ya era correcto, esta operación falla y se devuelve sin modificar. La segunda función aplica la primera a los nombres de las columnas y a todos los valores de texto del dataframe.

```python
def lectura_archivo(df_p):
    if not isinstance(df_p, str):
        return df_p
    for enc in ("cp1252", "latin-1"):
        try:
            return df_p.encode(enc).decode("utf-8")
        except (UnicodeEncodeError, UnicodeDecodeError):
            continue
    return df_p

def corregir_codificacio(df_p):
    df = df_p.copy()
    df.columns = [lectura_archivo(str(c)).strip() for c in df.columns]
    for col in df.select_dtypes(include="object").columns:
        df[col] = df[col].map(lectura_archivo)
        df[col] = df[col].str.strip()
    return df
```

Tras estas comprobaciones, ya se puede aplicar una limpieza del texto "basura" más precisa. También es necesario normalizar algunas etiquetas.

```python
import numpy as np

grups_valids = ["Grup A", "Grup B", "Grup C", "Grup D"]
df_limpieza_codificacion.loc[~df_limpieza_codificacion["Grup Professional"].isin(grups_valids), "Grup Professional"] = np.nan

def netejar_placeholders(df_p):
  df = df_p.copy()
  basura = {
      "Nom": ["???"],
      "Cognoms": ["###"],
      "DNI": ["XXXX"],
      "País d'origen": ["Narnia"],
      "Ciutat": ["Desconeguda"],
      "Càrrec": ["???"],
      "Salari mensual": ["molts"],
  }
  for col, valors in basura.items():
      df[col] = df[col].replace(valors, np.nan)
  return df

# Modificamos también las etiquetas de la columna de sexo.
mapa_genere = {
    "Dona": "Dona", "D": "Dona", "F": "Dona",
    "H": "Home", "Home": "Home", "M": "Home"}

df_limpieza_codificacion["Gènere"] = df_limpieza_codificacion["Gènere"].map(mapa_genere)

def eliminar_filas_irrecuperables(df_p, campos_clave=("Nom", "Cognoms", "DNI")):
  sin_identidad = df_p[list(campos_clave)].isna().all(axis=1)
  return df_p[~sin_identidad].reset_index(drop=True)
```

Esta función convierte a numéricas las columnas que tienen números guardados como texto (object).

Para cada columna, la función intenta convertirla con `pd.to_numeric(errors="coerce")`. Solo la convierte si todo su contenido es numérico (umbral de 1.0), así no se pierden datos legítimos sin darnos cuenta. Las columnas que ya son numéricas, las booleanas, las vacías y las que indiquemos en `excluir` (como el DNI) se saltan. Si todos los valores son enteros, se usa `Int64`, que admite nulos, y si no, `float`.

```python
def convertir_columnas_numericas(df_p, umbral=1.0, excluir=None, verbose=True):
  df = df_p.copy()
  excluir = set(excluir or [])
  convertidas, casi = [], []
  for col in df.columns:
      serie = df[col]
      if col in excluir or pd.api.types.is_numeric_dtype(serie) \
              or pd.api.types.is_bool_dtype(serie):
          continue
      con_dato = serie.notna()
      if con_dato.sum() == 0:
          continue
      if serie[con_dato].map(lambda v: isinstance(v, bool)).any():
          continue

      intento = pd.to_numeric(serie, errors="coerce")
      fallos = con_dato & intento.isna()
      proporcion = 1 - fallos.sum() / con_dato.sum()

      if proporcion >= umbral:
          if (intento.dropna() % 1 == 0).all():
              intento = intento.astype("Int64")
          df[col] = intento
          convertidas.append((col, str(intento.dtype), int(fallos.sum())))
      elif proporcion >= 0.8:
          ejemplos = serie[fallos].unique()[:5].tolist()
          casi.append((col, round(proporcion * 100, 1), ejemplos))
  return df
```

La siguiente función hace dos comprobaciones sobre el DataFrame: valida las fechas de nacimiento y revisa que el DNI sea único.

Primero recorre las columnas de día, mes y año de nacimiento y, si un valor está fuera de su rango (por ejemplo, un mes 99 o un año 3000), lo convierte en `NaN`. Solo se tocan los valores que tenían dato, para no contar como errores los nulos que ya existían. Los rangos se pasan a un diccionario.

Después se busca los DNIs repetidos y analiza cada grupo por separado, comparando los datos de identidad: nombre, apellidos, fecha de nacimiento y género. Si todas las filas del grupo tienen la misma identidad, se conserva la fila con menos nulos y se eliminan las demás. Si las identidades son distintas, pero el mismo DNI aparece asociado a personas diferentes, se descartan todas las filas del grupo.

```python
def validar_naixement_dni(df_p, rangos=None, col_dni="DNI", cols_identidad=None):
    df = df_p.copy()
    if rangos is None:
        rangos = {
            "Dia de Naixement": (1, 31),
            "Mes de Naixement": (1, 12),
            "Any de Naixement": (1940, 2010)}
    if cols_identidad is None:
        cols_identidad = ["Nom", "Cognoms", "Dia de Naixement",
                          "Mes de Naixement", "Any de Naixement", "Gènere"]

    # Rangs d'edat
    for col, (minimo, maximo) in rangos.items():
        fuera = df[col].notna() & ~df[col].between(minimo, maximo)
        df.loc[fuera, col] = np.nan

    # DNI's repetits
    con_dni = df[col_dni].notna()
    dups = df[con_dni & df.duplicated(col_dni, keep=False)].sort_values(col_dni)
    filas_a_eliminar = []
    for dni, grupo in dups.groupby(col_dni):
        n_identidades = grupo[cols_identidad].drop_duplicates().shape[0]
        if n_identidades == 1:
            mejor = grupo.isna().sum(axis=1).idxmin()
            filas_a_eliminar += [i for i in grupo.index if i != mejor]
        else:
            filas_a_eliminar += list(grupo.index)

    n_antes = len(df)
    df = df.drop(index=filas_a_eliminar).reset_index(drop=True)
    return df
```

Finalmente, se elabora la función main para integrarlas todas y poder sacar un dataframe medianamente limpio, con algunas comprobaciones hechas y algunas filas eliminadas. También se corrige el formato de algunas columnas, dejando el dataframe relativamente trabajado para el siguiente ejercicio.

```python
def neteja_main(df_p):
  df = promocion_encabezados(df_p.copy())
  df = lectura_archivo(df)
  df = corregir_codificacio(df)
  df = df.drop_duplicates().reset_index(drop=True)
  df = netejar_placeholders(df)
  grups_valids = ["Grup A", "Grup B", "Grup C", "Grup D"]
  df.loc[~df["Grup Professional"].isin(grups_valids), "Grup Professional"] = np.nan
  mapa_genere = {
      "Dona": "Dona", "D": "Dona", "F": "Dona",
      "H": "Home", "Home": "Home", "M": "Home"}
  df["Gènere"] = df["Gènere"].map(mapa_genere)
  df = eliminar_filas_irrecuperables(df)
  df = convertir_columnas_numericas(df, excluir=["DNI"])
  df = validar_naixement_dni(df)
  return df
```
