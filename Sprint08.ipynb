Siguiendo la lógica del sprint anterior, prepararé el limpiado de los datos con algunas funciones.

Esta primera coge el primer encabezado, situado en la primera fila, y los demás encabezados, situados en la tercera fila, y los guarda para reestructurar los nombres de los encabezados, eliminar los datos de las filas entre el encabezado antiguo y el nuevo y reestructurar los índices posteriormente.

También se añade un paso extra de limpieza de nombres para eliminar espacios.

```python
import pandas as pd
def promocion_encabezados(df_brut_p):
  df_brut_p = df_brut_p.iloc[:, 1:]
  df_brut_p.columns = df_brut_p.iloc[2].values
  df_brut_p.iloc[3:].reset_index(drop=True)
  df_brut_p = df_brut_p.iloc[3:].reset_index(drop=True)
  noves_columnes = []
  for col in df_brut_p.columns:
    col_str = str(col).strip()
    if "rrec" in col_str or "CÃ" in col_str:
      noves_columnes.append("Càrrec")
    else:
      col_neta = col_str.replace("_", " ")
      noves_columnes.append(col_neta)
  df_brut_p.columns = noves_columnes
  return df_brut_p
```
